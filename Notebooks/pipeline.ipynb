```python
import cv2
import matplotlib.pyplot as plt

from processing.pipeline import DocumentPipeline

pipeline = DocumentPipeline()

result = pipeline.process(
    "../uploads/sample2.png",
    save_outputs=True,
)

print(result["report"])

plt.figure(figsize=(8,10))
plt.imshow(cv2.cvtColor(result["original_image"], cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()

plt.figure(figsize=(8,10))
plt.imshow(result["clean_image"], cmap="gray")
plt.title("Clean Image")
plt.axis("off")
plt.show()

plt.figure(figsize=(8,10))
plt.imshow(cv2.cvtColor(result["boxed_image"], cv2.COLOR_BGR2RGB))
plt.title("OCR Detection")
plt.axis("off")
plt.show()

print("\nOCR TEXT\n")
print(result["ocr_text"])
```
