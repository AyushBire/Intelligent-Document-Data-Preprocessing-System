```python
import json
from processing.llm_processor import LLMProcessor

# Sample OCR-like text to test with — replace with real OCR output later
sample_text = """
INVOICE
Invoice No: INV-2026-0451
Date: 12/07/2026
Bill To: Rahul Sharma
Item: Web Hosting Plan - Annual
Amount: Rs. 4500.00
Total Due: Rs. 4500.00
"""

processor = LLMProcessor()
result = processor.process(sample_text)

print("---- RAW RESULT ----")
print(json.dumps(result, indent=4, ensure_ascii=False))

if "error" in result:
    print("\n⚠️  LLM call returned an error:", result["error"])
elif not result.get("structured_data", {}).get("dynamic_fields"):
    print("\n⚠️  structured_data is empty — check prompt/model response")
else:
    print("\n✅ LLM call succeeded with valid structured data.")
```
