```python
from processing.preprocessing import ImagePreprocessor
from processing.ocr import OCRProcessor
from processing.visualizer import OCRVisualizer
import cv2
import matplotlib.pyplot as plt

# Initialize modules
preprocessor = ImagePreprocessor()
ocr = OCRProcessor()
visualizer = OCRVisualizer()

# Load and preprocess
image = preprocessor.process("../uploads/sample_invoice.jpeg")

# Get OCR results
results = ocr.get_ocr_results(image)

# Draw bounding boxes
boxed = visualizer.draw_boxes(image, results)

# Display
plt.figure(figsize=(10, 12))
plt.imshow(cv2.cvtColor(boxed, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()

# Save
visualizer.save_image(
    boxed,
    "../outputs/boxed_invoice.png"
)
```
